# Chapitre 1 : Le Modèle Binomial de Cox-Ross-Rubinstein (CRR)

Le modèle de Cox-Ross-Rubinstein est un modèle à temps discret qui sert de fondation probabiliste pour approcher le modèle continu de Black-Scholes. Pour le construire rigoureusement, nous devons définir un espace probabilisé filtré complet.

On notera u (en référence à **Up**) et d (en référence à **Down**) pour modéliser une hausse (respectivement une baisse) relative de l'actif sur une période \(∆t\) donnée.

Par la suite, et par souci de simplification, la formalisation explicite de la filtration ainsi que la démonstration relative à la construction de l'espace probabilisable ne seront pas développées. 

---

## 1. L'espace fondamental et la théorie de la mesure

### 1.1. L'espace des états ($\Omega$)

Soit un horizon temporel fini $T > 0$ divisé en $N$ périodes identiques de longueur $\Delta t = \frac{T}{N}$.
À chaque période $n \in \{1, \dots, N\}$, l'actif sous-jacent subit une variation aléatoire modélisée par un facteur multiplicatif binaire :

* Soit une hausse $u > 1$ (état $u$).
* Soit une baisse $d = \frac{1}{u} < 1$ (état $d$).(on démontrera cela par la suite)

L'espace fondamental $\Omega$ est défini comme l'ensemble de toutes les trajectoires possibles (les réalisations) sur les $N$ périodes :

$$\Omega = \{u, d\}^N = \{(\omega_1, \omega_2, \dots, \omega_N) \mid \omega_i \in \{u, d\}, \forall i \in \{1, \dots, N\}\}$$

Cet ensemble est fini et son cardinal est $|\Omega| = 2^N$.

Notre objectif par la suite va être de déterminer des approximations asymptotiques de u et de d.

On appellera par la suite cet axiome, **l'axiome des rendements stationnaires**.

---

## 2. La mesure de probabilité historique ($\mathbb{P}$)

Sur l'espace mesurable $(\Omega, \mathcal{F})$, on définit une mesure de probabilité de référence $\mathbb{P}$ (la probabilité historique ou objective).

On pose que les variables aléatoires $(\zeta_i)_{1 \le i \le N}$ sont **indépendantes et identiquement distribuées (i.i.d.)** sous $\mathbb{P}$, suivant une loi de Bernoulli généralisée :

$$\mathbb{P}(\zeta_i = u) = p \quad \text{et} \quad \mathbb{P}(\zeta_i = d) = 1 - p$$

où $p \in \left]0, 1\right[$ représente la probabilité objective de hausse.

Par indépendance, la probabilité d'une trajectoire élémentaire $\omega = (\omega_1, \dots, \omega_N)$ est donnée par le produit des mesures de Bernoulli :

$$\mathbb{P}(\{\omega\}) = \prod_{i=1}^N \mathbb{P}(\zeta_i = \omega_i) = p^{k} (1-p)^{N-k}$$

où $k$ désigne le nombre de hausses (le nombre d'indices $i$ tels que $\omega_i = u$) dans la séquence $\omega$.

Le triplet $(\Omega, \mathcal{F}, \mathbb{P})$ forme un **espace probabilisé**.

## 3. Les axiomes du modèle CRR


### 3.1. L'axiome du non-arbitrage et son implication

#### Définition formelle d'un arbitrage :

Une **stratégie de portefeuille** auto-financée est un processus prévisible $\phi = (\Delta_n, \eta_n)_{n=1}^N$ où $\Delta_n$ représente le nombre d'actions détenues et $\eta_n$ le montant placé dans le compte d'épargne entre $n-1$ et $n$. La valeur du portefeuille à l'instant $n$ est notée $V_n^\phi$.

* **Définition (Arbitrage)** : Une opportunité d'arbitrage est une stratégie auto-financée $\phi$ telle que :

1. $V_0^\phi = 0$ (aucun investissement initial).
2. $V_N^\phi \ge 0$ presque sûrement sous $\mathbb{P}$ ($\mathbb{P}(V_N^\phi \ge 0) = 1$).
3. $\mathbb{P}(V_N^\phi > 0) > 0$ (possibilité stricte de gain positif).

Autrement dit, c'est un portefeuille qui permet de gagner de l'argent sans aucun risque de perte et sans mise initiale.

#### Conséquence direct du principe de non-arbitrage :

On se place dans un modèle à une période (de $0$ à $1$). L'actif risqué a un prix initial $S_0$ et peut valoir à l'instant $1$ soit $S_1 = u S_0$, soit $S_1 = d S_0$ (avec $u > d > 0$). L'actif sans risque croît au taux $r$, de sorte qu'un euro investi en $0$ vaut $1 + r$ en $1$ (on pose ainsi $B_0 = 1$ et $B_1 = 1 + r$).

On souhaite montrer que l'hypothèse d'**absence d'arbitrage** impose nécessairement l'inégalité stricte :

$$d < 1 + r < u$$

#### Raisonnement par l'absurde

Supposons que cette condition ne soit pas respectée. Il y a deux cas de violation possibles, que nous allons rejeter un à un en construisant explicitement une opportunité d'arbitrage.

---

##### Cas 1 : Supposons que $1 + r \le d$

Si le taux sans risque est inférieur ou égal au plus mauvais rendement possible de l'actif risqué, construisons le portefeuille suivant en $0$ :

* On achète $1$ action : $\Delta = 1$
* On emprunte la somme $S_0$ au taux sans risque : $\eta = -S_0$

La valeur initiale de ce portefeuille est :

$$V_0 = \Delta S_0 + \eta B_0 = 1 \cdot S_0 - S_0 \cdot 1 = 0$$

*(Aucun investissement initial, condition 1 d'un arbitrage).*

À l'instant $1$, la valeur du portefeuille devient :

$$V_1 = S_1 - S_0(1 + r)$$

Puisque $S_1$ peut valoir $u S_0$ ou $d S_0$, regardons la valeur de $V_1$ dans le pire des cas (quand $S_1 = d S_0$) :

$$V_1 \ge d S_0 - S_0(1 + r) = S_0 \left( d - (1 + r) \right)$$

Par hypothèse dans ce cas, $1 + r \le d$, ce qui signifie que $d - (1 + r) \ge 0$.
De plus, en cas de hausse ($S_1 = u S_0$), comme $u > d \ge 1 + r$, le gain est strictement positif.
On a donc $\mathbb{P}(V_1 \ge 0) = 1$ et $\mathbb{P}(V_1 > 0) > 0$.

Cela constitue une **opportunité d'arbitrage**, ce qui contredit l'hypothèse de départ du modèle. Par conséquent, l'inégalité $1 + r \le d$ est impossible, donc **$1 + r > d$**.

---

##### Cas 2 : Supposons que $1 + r \ge u$

Si le taux sans risque est supérieur ou égal au meilleur rendement possible de l'action, construisons le portefeuille symétrique :

* On vend à découvert $1$ action : $\Delta = -1$
* On place la somme $S_0$ au taux sans risque à la banque : $\eta = S_0$

La valeur initiale du portefeuille est :

$$V_0 = -1 \cdot S_0 + 1 \cdot S_0 = 0$$

À l'instant $1$, la valeur du portefeuille est :

$$V_1 = S_0(1 + r) - S_1$$

Dans le pire des cas pour notre position (quand l'action monte au maximum, $S_1 = u S_0$) :

$$V_1 \ge S_0(1 + r) - u S_0 = S_0 \left( (1 + r) - u \right)$$

Puisque l'on a supposé $1 + r \ge u$, cette quantité est positive ou nulle, et elle est strictement positive en cas de baisse ($S_1 = d S_0$).
On obtient à nouveau une **opportunité d'arbitrage**.

Par conséquent, l'inégalité $1 + r \ge u$ est également impossible, donc **$1 + r < u$**.

---

#### Conclusion

En combinant les deux résultats obtenus par l'absurde ($d < 1 + r$ et $1 + r < u$), on en conclut que l'absence d'arbitrage implique nécessairement :

$$d < 1 + r < u$$

### 3.2. L'axiome de neutralité face au risque et son implication

L'introduction de la mesure de probabilité risque-neutre $\mathbb{Q}$ repose sur un axiome fondamental : **l'axiome de neutralité face au risque**.

* **Définition de l'axiome :**
Dans le monde risque-neutre, les investisseurs ne requièrent **aucune prime de risque** pour accepter l'incertitude liée à la détention d'un actif risqué. Par conséquent, le rendement espéré de n'importe quel actif (risqué ou non) est censé être égal au taux d'intérêt sans risque.

#### Implication sur l'espérance des facteurs de variation

Rappelons que les variables aléatoires $(\zeta_k)_{1 \le k \le N}$ représentent les facteurs multiplicateurs de l'actif entre deux périodes ($u$ ou $d$).

Dans le cadre de la modélisation binomiale, on pose que l'espérance mathématique du facteur de variation de l'actif sur une période de longueur $\Delta t$ reflète la croissance associée au taux d'intérêt sans risque.

Dans un premier temps, avec un **taux d'intérêt discret** $r_{disc}$ proportionnel à la période, cette espérance s'écrit :

$$\mathbb{E}[\zeta_k] = p \cdot u + (1 - p) \cdot d = 1 + r_{disc} \Delta t$$

#### L'équivalence avec le taux d'intérêt continu

Bien que le taux discret soit intuitif, il s'avère lourd à manipuler lors du passage à la limite vers le modèle continu. En finance mathématique, on lui préfère la capitalisation continue, modélisée par une exponentielle $e^{r \Delta t}$ (où $r$ est le taux sans risque continu).

Il existe une relation mathématique **exacte** pour passer d'un taux discret à un taux continu équivalent sur la période $\Delta t$ :

$$1 + r_{disc} \Delta t = e^{r \Delta t}$$

En appliquant le logarithme népérien, on obtient la formule de conversion précise :

$$r = \frac{1}{\Delta t} \ln(1 + r_{disc} \Delta t)$$

#### Le choix du modèle continu

Pour la suite de notre construction vers le modèle de Black-Scholes, **nous poserons désormais directement que le facteur de croissance est sous forme exponentielle**, en notant $r$ le taux sans risque continu :

$$\mathbb{E}[\zeta_k] = p \cdot u + (1 - p) \cdot d = e^{r \Delta t}$$

Cette écriture exponentielle est un choix fondamental : elle simplifie considérablement les calculs algébriques et permettra d'appliquer naturellement le Théorème Central Limite pour obtenir la loi log-normale de l'actif à l'échéance. (Cela va être clair avec l'utilisation des développements limités par la suite.

---


### 4. Détermination des facteurs $u$ et $d$

La question à se poser maintenant est :

> Quelles formules algébriques donner à $u$ et $d$ pour rester en adéquation avec le modèle théorique que l'on vient de créer ?

#### Définition du problème

$(\zeta_k)_{1 \le k \le N}$ est la famille correspondant à la valorisation de l'actif $A$ sur un pas de temps $\Delta t = \frac{T}{N}$.

Ces variables sont indépendantes et identiquement distribuées (**i.i.d.**) et régies par la table suivante :

| Valeur de $\zeta_k$ | $u$ (hausse) | $d$ (baisse) |
| --- | --- | --- |
| **Probabilité** | $p$ | $1 - p$ |

Par hypothèse sur le taux sans risque continu, on a :

$$\mathbb{E}[\zeta_k] = p \cdot u + (1 - p) \cdot d = e^{r \Delta t}$$

Le prix de l'actif risqué à l'instant $T$ ($S_T$) s'exprime par le produit des facteurs successifs :

$$S_T = S_0 \cdot \prod_{k=1}^{N} \zeta_k = S_0 \cdot \zeta_1 \cdot \zeta_2 \cdots \zeta_N$$

#### Propriété de log-additivité et passage aux variables $Y_k$

En utilisant la propriété du logarithme, on se ramène à une somme :

$$\ln(S_T) = \ln(S_0) + \sum_{k=1}^{N} \ln(\zeta_k)$$

Posons ces nouvelles variables aléatoires i.i.d. $Y_k$ :

$$Y_k = \ln(\zeta_k) \quad \text{pour tout } k \in \{1, \dots, N\}$$

Notons $\mu_Y = \mathbb{E}[Y_k]$ l'espérance et $\sigma_Y^2 = \mathbb{V}[Y_k]$ la variance de ces variables sur un pas de temps.

#### Espérance de $\ln(S_T)$

Par linéarité de l'espérance et indépendance, l'espérance du logarithme de l'actif à l'échéance s'écrit :

$$\mathbb{E}[\ln(S_T)] = \ln(S_0) + \sum_{k=1}^{N} \mathbb{E}[Y_k] = \ln(S_0) + N \cdot \mu_Y$$

#### Variance de $\ln(S_T)$ et lien avec la volatilité $\sigma$

Par les propriétés de la variance (la variance d'une constante est nulle et les variances s'additionnent pour des variables indépendantes) :

$$\mathbb{V}[\ln(S_T)] = \sum_{k=1}^{N} \mathbb{V}[Y_k] = N \cdot \sigma_Y^2$$

Puisque le nombre total de pas est lié au temps total par $N = \frac{T}{\Delta t}$, on peut réécrire cette variance sous la forme :

$$\mathbb{V}[\ln(S_T)] = \left(\frac{T}{\Delta t}\right) \cdot \sigma_Y^2 = T \cdot \left(\frac{\sigma_Y^2}{\Delta t}\right)$$

Quand on fait tendre le nombre de pas $N$ vers l'infini (et donc que $\Delta t$ tend vers $0$), le Théorème Central Limite assure que la somme $\sum Y_k$ converge vers une loi normale. Pour que la variance totale de cette somme à l'horizon fixe $T$ reste bien proportionnelle au temps (une exigence fondamentale de stabilité pour modéliser un actif financier), le rapport $\frac{\sigma_Y^2}{\Delta t}$ doit être égal à une constante positive que l'on note $\sigma^2$ (la variance instantanée ou volatilité globale au carré) :

$$\frac{\sigma_Y^2}{\Delta t} = \sigma^2 \quad \implies \quad \sigma_Y^2 = \sigma^2 \Delta t$$

---


Maintenant que nous avons établi que $\sigma_Y^2 = \sigma^2 \Delta t$, nous voulons exprimer cette variance directement à partir des paramètres de notre modèle discret ($u$, $d$ et $p$).

#### Le théorème de König-Huygens

Pour calculer la variance de la variable aléatoire $Y_k = \ln(\zeta_k)$, on utilise la formule de **König-Huygens** (qui relie la variance à l'espérance et à l'espérance du carré) :

$$\sigma_Y^2 = \mathbb{V}[Y_k] = \mathbb{E}[Y_k^2] - (\mathbb{E}[Y_k])^2$$

Rappelons la loi de probabilité de $Y_k$ :

* $Y_k = \ln(u)$ avec la probabilité $p$
* $Y_k = \ln(d)$ avec la probabilité $1 - p$

À partir de cette loi, on calcule explicitement l'espérance et l'espérance du carré :

* $\mathbb{E}[Y_k] = p \cdot \ln(u) + (1 - p) \cdot \ln(d)$
* $\mathbb{E}[Y_k^2] = p \cdot (\ln(u))^2 + (1 - p) \cdot (\ln(d))^2$

En injectant ces expressions dans la formule de König-Huygens, on obtient :

$$\sigma_Y^2 = p \cdot (\ln(u))^2 + (1 - p) \cdot (\ln(d))^2 - \left[ p \cdot \ln(u) + (1 - p) \cdot \ln(d) \right]^2$$

---

#### Développement algébrique et identités remarquables

Pour simplifier cette expression, on développe le carré du second terme à l'aide de l'identité remarquable $(a + b)^2 = a^2 + 2ab + b^2$ :

$$\sigma_Y^2 = p (\ln(u))^2 + (1 - p) (\ln(d))^2 - \left[ p^2 (\ln(u))^2 + 2p(1-p)\ln(u)\ln(d) + (1-p)^2 (\ln(d))^2 \right]$$

En regroupant les termes en $(\ln(u))^2$ et en $(\ln(d))^2$ :

$$\sigma_Y^2 = (\ln(u))^2 \cdot [p - p^2] + (\ln(d))^2 \cdot [(1-p) - (1-p)^2] - 2p(1-p)\ln(u)\ln(d)$$

En factorisant par $p$ dans le premier terme et $(1-p)$ dans le second, on remarque que $p - p^2 = p(1-p)$ et $(1-p) - (1-p)^2 = (1-p)p = p(1-p)$. On peut donc tout factoriser par le terme commun $p(1-p)$ :

$$\sigma_Y^2 = p(1-p) \left[ (\ln(u))^2 - 2\ln(u)\ln(d) + (\ln(d))^2 \right]$$

L'expression entre crochets est une **seconde identité remarquable**, qui correspond au développement de $(a - b)^2 = a^2 - 2ab + b^2$ (avec $a = \ln(u)$ et $b = \ln(d)$). On obtient ainsi :

$$\sigma_Y^2 = p(1-p) \left[ \ln(u) - \ln(d) \right]^2$$

En utilisant la propriété classique du logarithme $\ln(u) - \ln(d) = \ln\left(\frac{u}{d}\right)$, on arrive à la forme finale de la variance :

$$\sigma_Y^2 = p(1-p) \left[ \ln\left(\frac{u}{d}\right) \right]^2$$

---

#### Le lien final avec la volatilité

Puisque nous avons établi précédemment que $\sigma_Y^2 = \sigma^2 \Delta t$, on obtient l'équation clé de la variance pour notre modèle :

$$p(1-p) \left[ \ln\left(\frac{u}{d}\right) \right]^2 = \sigma^2 \Delta t$$

En prenant la racine carrée de l'équation de la variance, on obtient :

$$\sqrt{p(1-p)} \ln\left(\frac{u}{d}\right) = \sigma \sqrt{\Delta t}$$

Pour simplifier le modèle et assurer la recombinaison de l'arbre binomial (c'est-à-dire qu'une hausse suivie d'une baisse ramène l'actif à son niveau initial), on pose l'hypothèse que :

$$S_{k+2} = S_k \cdot u \cdot d = S_k$$

Puisque $S_k \neq 0$ (car un actif dont le prix s'annule perd toute valeur et sort du marché, et cela n'a aucun sens d'être analysé dans le cadre de notre modèle), cela implique directement :

$$u \cdot d = 1 \quad \implies \quad d = \frac{1}{u} = u^{-1}$$

*(Le facteur de baisse $d$ est l'inverse, ou le "symétrique", du facteur de hausse $u$).*

En réinjectant cette condition dans notre équation, on simplifie le logarithme :

$$\sqrt{p(1-p)} \ln\left(\frac{u}{u^{-1}}\right) = \sqrt{p(1-p)} \ln(u^2) = 2\sqrt{p(1-p)} \ln(u) = \sigma \sqrt{\Delta t}$$

---

#### Comportement asymptotique de $p$ et de $u$ lorsque $\Delta t \to 0$

À partir de l'équation de l'espérance du facteur de variation, $p \cdot u + (1 - p) \cdot d = e^{r \Delta t}$, exprimons explicitement la probabilité $p$ :

$$p(u - d) + d = e^{r \Delta t} \implies p = \frac{e^{r \Delta t} - d}{u - d}$$

Lorsque le pas de temps $\Delta t$ tend vers $0$, l'intervalle devient infiniment court, ce qui signifie que le prix de l'actif n'a pas le temps de varier sur une période élémentaire. En effet :

$$\lim_{\Delta t \to 0} \zeta_k = 1$$

Par conséquent, les facteurs de variation $u$ et $d$ convergent tous deux vers $1$ :

* Pour la hausse : $u \to 1$
* Pour la baisse : $d \to 1$


---

#### Développements limités (DL) de $u$, $d$ et de leurs logarithmes

Pour analyser finement le comportement des paramètres lorsque le pas de temps $\Delta t \to 0$, posons un nouveau paramètre de petitesse $h$ défini par la racine carrée du pas de temps :

$$h = \sqrt{\Delta t} \quad \implies \quad \Delta t = h^2$$

Puisque $u \to 1$ et $d \to 1$ quand $\Delta t \to 0$, on exprime $u$ au voisinage de $1$ à l'aide d'un développement limité (via le théorème de Taylor-Young) :

$$u = 1 + c \cdot h + o(h)$$

**Justification de la condition $c > 0$ :**

Par définition, un facteur de hausse ($u$) représente une augmentation de la valeur de l'actif sur le pas de temps élémentaire, ce qui implique que $u > 1$, soit $u - 1 > 0$. En injectant son développement limité, on a $u - 1 = c \cdot h + o(h) > 0$. Puisque $h = \sqrt{\Delta t} > 0$, le coefficient directeur $c$ doit nécessairement être strictement positif ($c > 0$), faute de quoi une "hausse" entrainerait une stagnation ou une baisse du prix.

Grâce à la condition de symétrie $u \cdot d = 1$ (soit $d = \frac{1}{u}$), on en déduit directement le développement de $d$ :

$$d = \frac{1}{1 + c \cdot h + o(h)} = 1 - c \cdot h + o(h)$$

#### Application directe du développement limité du logarithme

Pour obtenir le logarithme de ces facteurs, on utilise le développement limité usuel au voisinage de $0$ :

$$\ln(1 + x) = x + o(x)$$

* **Pour le facteur de hausse $u$ :**

$$\ln(u) = \ln(1 + c \cdot h + o(h)) = c \cdot h + o(h)$$

* **Pour le facteur de baisse $d$ :**

$$\ln(d) = \ln(1 - c \cdot h + o(h)) = -c \cdot h + o(h)$$

À partir de la relation sur les logarithmes que nous venons d'établir, $\ln(u) = c \cdot h + o(h)$, on peut déduire les développements limités (DL) de $u$ et $d$ à l'ordre 2 en appliquant la fonction exponentielle et son développement usuel ($e^x = 1 + x + \frac{x^2}{2} + o(x^2)$) :

* **Pour le facteur de hausse $u$ :**

$$u = e^{\ln(u)} = e^{c \cdot h + o(h)} = 1 + c \cdot h + \frac{c^2}{2} h^2 + o(h^2)$$

* **Pour le facteur de baisse $d$ :**
Puisque $\ln(d) = -c \cdot h + o(h)$ (ou par la relation de symétrie $d = \frac{1}{u}$), on obtient de même :

$$d = e^{\ln(d)} = e^{-c \cdot h + o(h)} = 1 - c \cdot h + \frac{c^2}{2} h^2 + o(h^2)$$

En rappelant que le pas de temps est relié à notre paramètre de petitesse par $\Delta t = h^2$, on cherche le développement limité de $e^{r \Delta t}$ au voisinage de $0$ (lorsque $h \to 0$).

En utilisant le développement limité usuel de la fonction exponentielle $e^x = 1 + x + o(x)$ :

$$e^{r \Delta t} = e^{r h^2} = 1 + r h^2 + o(h^2)$$

Déterminons les développements limités du numérateur et du dénominateur en fonction de $h$ (avec $\Delta t = h^2$).

---

#### 1. Développement limité du numérateur : $e^{r \Delta t} - d$
On calcule la différence :

$$e^{r \Delta t} - d = \left(1 + r h^2 + o(h^2)\right) - \left(1 - c h + \frac{c^2}{2} h^2 + o(h^2)\right)$$

En simplifiant les termes en $1$ :

$$e^{r \Delta t} - d = c h + r h^2 - \frac{c^2}{2} h^2 + o(h^2)$$

En factorisant par $h$:

$$e^{r \Delta t} - d = h \left[ c + r h - \frac{c^2}{2} h + o(h) \right] = h \left[ c + \left( r - \frac{c^2}{2} \right) h + o(h) \right]$$

---

#### 2. Développement limité du dénominateur : $u - d$
On calcule la différence :

$$u - d = \left(1 + c h + \frac{c^2}{2} h^2 + o(h^2)\right) - \left(1 - c h + \frac{c^2}{2} h^2 + o(h^2)\right)$$

Les termes en $1$ et en $h^2$ s'annulent parfaitement :

$$u - d = 2 c h + o(h^2) = h \left[ 2c + o(1) \right]$$

#### Développement limité et limite de la probabilité $p$

En remplaçant le numérateur et le dénominateur par leurs développements limités respectifs en fonction de $h$ (avec $\Delta t = h^2$), l'expression de la probabilité neutre au risque $p$ devient :

$$p = \frac{e^{r \Delta t} - d}{u - d} = \frac{h \left[ c + \left( r - \frac{c^2}{2} \right) h + o(h) \right]}{h \left[ 2c + o(1) \right]}$$

---

#### Simplification algébrique

Puisque $h \neq 0$, on peut simplifier la fraction en factorisant et en divisant le numérateur et le dénominateur par $h$ :

$$p = \frac{c + \left( r - \frac{c^2}{2} \right) h + o(h)}{2c + o(1)}$$

---

#### Passage à la limite lorsque $h \to 0$ (ou $\Delta t \to 0$)

Lorsque le pas de temps tend vers $0$ ($h \to 0$), les termes en $h$ au numérateur tendent vers $0$, et le terme $o(1)$ au dénominateur tend vers $0$.

Par application des règles de continuité et de quotient de limites d'une fonction d'un infiniment petit, on obtient la limite de $p$ :

$$\lim_{h \to 0} p = \frac{c}{2c} = \frac{1}{2}$$

---

#### Développement limité de $p$ en fonction de $1/2$

En reprenant l'expression de la probabilité $p$ obtenue précédemment avec $h = \sqrt{\Delta t}$, on peut l'écrire sous la forme d'un développement limité au voisinage de $1/2$ :

$$p = \frac{1}{2} + o(\sqrt{\Delta t})$$

Par conséquent, pour le terme complémentaire $1 - p$, on a :

$$1 - p = 1 - \left( \frac{1}{2} + o(\sqrt{\Delta t}) \right) = \frac{1}{2} + o(\sqrt{\Delta t})$$

---

#### Expression de $\sqrt{p(1-p)}$

En multipliant $p$ et $1-p$ pour l'insérer dans notre formule de variance :

$$p(1-p) = \left( \frac{1}{2} + o(\sqrt{\Delta t}) \right) \left( \frac{1}{2} + o(\sqrt{\Delta t}) \right) = \frac{1}{4} + o(\sqrt{\Delta t})$$

Pour appliquer proprement le développement limité de la fonction racine carrée $\sqrt{1+x} = 1 + \frac{1}{2}x + o(x)$ au voisinage de $0$, on factorise d'abord par $\frac{1}{4}$ à l'intérieur de la racine :

$$\sqrt{p(1-p)} = \sqrt{\frac{1}{4} \left( 1 + o(\sqrt{\Delta t}) \right)}$$

En sortant le $\frac{1}{4}$ de la racine (ce qui donne $\frac{1}{2}$) :

$$\sqrt{p(1-p)} = \frac{1}{2} \sqrt{1 + o(\sqrt{\Delta t})}$$

En appliquant le développement limité de $\sqrt{1+X} = 1 + o(X)$ :

$$\sqrt{p(1-p)} = \frac{1}{2} \left( 1 + o(\sqrt{\Delta t}) \right) = \frac{1}{2} + o(\sqrt{\Delta t})$$

---

#### Déduction de la forme de $\ln(u)$

En réinjectant cette expression dans l'équation de la variance simplifiée ($2\sqrt{p(1-p)} \ln(u) = \sigma \sqrt{\Delta t}$) :

$$2 \left( \frac{1}{2} + o(\sqrt{\Delta t}) \right) \ln(u) = \sigma \sqrt{\Delta t}$$

Ce qui se simplifie en :

$$(1 + o(\sqrt{\Delta t})) \ln(u) = \sigma \sqrt{\Delta t}$$

En isolant $\ln(u)$ par division :

$$\ln(u) = \frac{\sigma \sqrt{\Delta t}}{1 + o(\sqrt{\Delta t})}$$

En utilisant le développement limité de l'inverse $\frac{1}{1 + x} = 1 - x + o(x)$, on obtient finalement :

$$\ln(u) = \sigma \sqrt{\Delta t} \left( 1 + o(\sqrt{\Delta t}) \right) = \sigma \sqrt{\Delta t} + o(\sqrt{\Delta t})$$

---

#### 1. Équivalent du facteur de hausse $u$

On sait que :

$$\ln(u) = \sigma \sqrt{\Delta t} + o(\sqrt{\Delta t})$$

Pour trouver un équivalent de $u$, revenons à sa définition via l'exponentielle :

$$u = e^{\ln(u)} = e^{\sigma \sqrt{\Delta t} + o(\sqrt{\Delta t})}$$

Grâce aux propriétés algébriques de l'exponentielle, on sépare les termes :

$$u = e^{\sigma \sqrt{\Delta t}} \cdot e^{o(\sqrt{\Delta t})}$$

Vérifions la définition de l'équivalence en étudiant la limite du quotient lorsque $\Delta t \to 0$ :

$$\frac{u}{e^{\sigma \sqrt{\Delta t}}} = e^{o(\sqrt{\Delta t})}$$

Puisque $\lim_{\Delta t \to 0} o(\sqrt{\Delta t}) = 0$ et que la fonction exponentielle est continue en $0$ ($e^{0} = 1$) :

$$\lim_{\Delta t \to 0} e^{o(\sqrt{\Delta t})} = 1$$

La limite du quotient étant égale à $1$, on a par définition l'équivalent recherché :

$$u \underset{\Delta t \to 0}{\sim} e^{\sigma \sqrt{\Delta t}}$$

---

#### 2. Équivalent du facteur de baisse $d$

Grâce à la relation de recombinaison et de symétrie de l'arbre démontrée plus tôt ($u \cdot d = 1$), on a :

$$d = \frac{1}{u}$$

En utilisant les règles sur les équivalents d'inverses :

$$d \underset{\Delta t \to 0}{\sim} \frac{1}{e^{\sigma \sqrt{\Delta t}}}$$

Soit sous forme exponentielle :

$$d \underset{\Delta t \to 0}{\sim} e^{-\sigma \sqrt{\Delta t}}$$


Ainsi, dans le modèle de Black-Scholes, on posera par définition $u = e^{\sigma \sqrt{\Delta t}}$ et $d = e^{-\sigma \sqrt{\Delta t}}$ à partir de ces équivalents : le modèle discret ne fournit donc que des estimations asymptotiques et non les valeurs exactes du modèle continu.

---

### Convergence vers la loi log-normale par le Théorème de la Limite Centrale (TCL)

L'objectif final est de démontrer que, lorsque le nombre de pas de temps $N$ tend vers l'infini (ce qui correspond à un pas élémentaire $\Delta t$ tendant vers $0$), la distribution discrète du prix de l'actif converge vers la loi log-normale que l'on réutilisera ensuite dans le modèle de Black-Scholes.

---

#### 1. Développement de la probabilité $p$ et espérance d'un pas

Reprenons l'expression analytique de la probabilité neutre au risque $p$ sous forme de fraction développée :

$$p = \frac{c + \left( r - \frac{c^2}{2} \right) h + o(h)}{2c + o(1)}$$

En posant $c = \sigma$ (le paramètre de volatilité identifié lors de notre calibration), on obtient par division le développement limité de $p$ à l'ordre 1 :

$$p = \frac{1}{2} + \frac{1}{2} \left( \frac{r}{\sigma} - \frac{\sigma}{2} \right) h + o(h)$$

Le log-rendement sur un pas élémentaire est la variable aléatoire $Y_k = \ln(\zeta_k)$, dont l'espérance s'exprime par :

$$E(Y_k) = p \ln(u) + (1-p) \ln(d)$$

En remplaçant $\ln(u) = \sigma h + o(h)$ et $\ln(d) = -\sigma h + o(h)$, on factorise :

$$E(Y_k) = (2p - 1)\sigma h + o(h)$$

En réinjectant le développement de $2p - 1$ tiré du DL de $p$ :

$$E(Y_k) = \left[ \left( \frac{r}{\sigma} - \frac{\sigma}{2} \right) h + o(h) \right] \sigma h + o(h) = \left( r - \frac{\sigma^2}{2} \right) h^2 + o(h^2)$$

Puisque le pas temporel élémentaire est défini par $h^2 = \Delta t$, on obtient l'espérance sur un pas :

$$E(Y_k) = \left( r - \frac{\sigma^2}{2} \right) \Delta t + o(\Delta t)$$

---

#### 2. Calcul rigoureux de la variance sur un pas $V(Y_k)$

En utilisant la formule exacte de la variance trouvée plus tôt:

$$V(Y_k) = p(1-p) \left[ \ln\left(\frac{u}{d}\right) \right]^2$$

* **Évaluation du terme $\ln(u/d)$ :**

$$\ln\left(\frac{u}{d}\right) = \ln(u) - \ln(d) = (\sigma h + o(h)) - (-\sigma h + o(h)) = 2\sigma h + o(h)$$

En élevant au carré :

$$\left[ \ln\left(\frac{u}{d}\right) \right]^2 = \left( 2\sigma h + o(h) \right)^2 = 4\sigma^2 h^2 + o(h^2) = 4\sigma^2 \Delta t + o(\Delta t)$$

* **Produit des probabilités $p(1-p)$ :**
Puisque $p = \frac{1}{2} + o(1)$, on a $p(1-p) = \frac{1}{4} + o(1)$.
* **Combinaison :**

$$V(Y_k) = \left( \frac{1}{4} + o(1) \right) \left( 4\sigma^2 \Delta t + o(\Delta t) \right) = \sigma^2 \Delta t + o(\Delta t)$$

---

#### 3. Passage à l'horizon global $T$ et application rigoureuse du TCL

Le log-rendement global entre l'instant initial $0$ et l'échéance finale $T$ (où $T = N \Delta t$) est la somme de $N$ variables aléatoires $Y_k$ indépendantes et identiquement distribuées (i.i.d.) :

$$\ln\left(\frac{S_T}{S_0}\right) = \sum_{k=1}^N Y_k$$

* **Espérance globale :**
En sommant l'espérance de chaque pas et en gérant rigoureusement le terme d'erreur :

$$E\left(\ln\left(\frac{S_T}{S_0}\right)\right) = \sum_{k=1}^N \left[ \left( r - \frac{\sigma^2}{2} \right) \Delta t + o(\Delta t) \right] = \left( r - \frac{\sigma^2}{2} \right) N \Delta t + N \cdot o(\Delta t)$$

Puisque $N \Delta t = T$ et que $N \cdot o(\Delta t) = \frac{T}{\Delta t} o(\Delta t) = o(1)$ (terme tendant vers $0$ quand $\Delta t \to 0$), il vient à la limite :

$$E\left(\ln\left(\frac{S_T}{S_0}\right)\right) \underset{\Delta t \to 0}{\longrightarrow} \left( r - \frac{\sigma^2}{2} \right) T$$

* **Variance globale :**
De la même manière, en sommant la variance de chaque pas :

$$V\left(\ln\left(\frac{S_T}{S_0}\right)\right) = \sum_{k=1}^N \left[ \sigma^2 \Delta t + o(\Delta t) \right] = \sigma^2 N \Delta t + N \cdot o(\Delta t) = \sigma^2 T + o(1)$$

À la limite lorsque $\Delta t \to 0$ (soit $N \to +\infty$), le terme d'erreur $o(1)$ s'annule pour donner exactement $\sigma^2 T$.

---

#### Conclusion

Les variables $Y_k$ étant i.i.d. et possédant des moments d'ordre 2 finis, le **Théorème de la Limite Centrale (TCL)** garantit que lorsque $N \to +\infty$ (soit $\Delta t \to 0$), la somme des log-rendements converge en loi vers une distribution normale :

$$\ln\left(\frac{S_N}{S_0}\right) \underset{N \to +\infty}{\xrightarrow{\text{Loi}}} \mathcal{N}\left(\left(r - \frac{\sigma^2}{2}\right)T, \, \sigma^2 T\right)$$

---

#### 4. Loi de $\ln(S_T)$ et expression en distribution

Puisque le log-rendement global est défini par $\ln\left(\frac{S_T}{S_0}\right) = \ln(S_T) - \ln(S_0)$, on en déduit directement l'expression de $\ln(S_T)$ :

$$\ln(S_T) = \ln(S_0) + \ln\left(\frac{S_N}{S_0}\right)$$

Par les propriétés de la loi normale, l'ajout d'une constante déterministe $\ln(S_0)$ à une variable suivant une loi normale ne fait que décaler son espérance. Par conséquent, lorsque $\Delta t \to 0$ (soit $N \to +\infty$), la variable $\ln(S_T)$ converge en loi vers :

$$\ln(S_T) \underset{N \to +\infty}{\xrightarrow{\text{Loi}}} \mathcal{N}\left(\ln(S_0) + \left(r - \frac{\sigma^2}{2}\right)T, \, \sigma^2 T\right)$$

Pour exprimer cela de manière explicite à l'aide d'une variable aléatoire gaussienne standard $Z \sim \mathcal{N}(0, 1)$, toute variable normale peut se réécrire sous la forme de son espérance plus l'écart-type multiplié par $Z$.

On obtient ainsi l'égalité en distribution (ou égalité en loi) suivante :

$$\ln(S_T) \overset{\mathcal{L}}{=} \ln(S_0) + \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} \, Z$$

où $Z$ désigne une variable aléatoire suivant la loi normale centrée réduite $\mathcal{N}(0, 1)$.

#### 5. Déduction de la loi de $S_T$ et passage au modèle de Black-Scholes

En appliquant la fonction exponentielle aux deux membres de l'égalité en distribution de $\ln(S_T)$, on obtient directement l'égalité en loi pour le prix de l'actif à l'échéance $S_T$ :

$$S_T \overset{\mathcal{L}}{=} S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} \, Z \right)$$

où $Z \sim \mathcal{N}(0, 1)$.

Par définition, une variable aléatoire dont le logarithme suit une loi normale est une variable qui suit une **loi log-normale**. Nous venons ainsi de prouver rigoureusement que le modèle discret converge en loi vers la structure de marché continue requise.

---

#### Utilisation pratique dans le modèle de Black-Scholes

Dans la pratique du modèle de Black-Scholes, on effectuera tous les calculs d'évaluation et d'espérances en passant par **$\ln\left(\frac{S_T}{S_0}\right)$** plutôt que directement par $S_T$.

En effet, travailler avec le logarithme du prix revient à manipuler des variables gaussiennes, ce qui simplifie énormément les intégrations, le calcul des espérances conditionnelles et l'évaluation des options, les propriétés de la loi normale étant beaucoup plus aisées à manier algébriquement que celles de la loi log-normale.

# Chapitre 2: La formule de Black-Shole

## Le cas du call Européen.

### Étape 1 : Qu'est-ce qu'un « Payoff » ?

#### 1. Définition intuitive

Dans le domaine des produits dérivés, un **payoff** (ou fonction de gain / flux terminal) représente la « règle du jeu » financière fixée par le contrat. C'est la valeur que vaudra l'option **à la date d'échéance $T$**, en fonction uniquement du prix de l'actif sous-jacent à ce moment-là, noté $S_T$.

#### 2. Focus sur le Call Européen

Un **Call européen** confère à son détenteur le *droit* (mais non l'obligation) d'acheter un actif sous-jacent à un prix fixé à l'avance, appelé **prix d'exercice** ou *strike* $K$, à une date précise $T$.

Regardons ce qui se passe à l'échéance $T$ selon la valeur du prix de l'actif $S_T$ :

* **Cas 1 : L'actif vaut plus que le strike ($S_T > K$)**:
Nous avons le droit d'acheter l'actif au prix bas $K$. Comme il vaut $S_T$ sur le marché, nous exerçons notre option : nous l'achetons à $K$ et nous pouvons (mentalement ou réellement) le revendre immédiatement au prix du marché $S_T$. Notre gain net est donc de :* $S_T - K$.

* **Cas 2 : L'actif vaut moins ou autant que le strike ($S_T \le K$)**
Il est absurde d'exercer notre droit d'acheter l'actif à $K$ alors qu'il s'achète moins cher sur le marché libre. Nous laissons l'option expirer sans l'exercer. Notre gain est donc de : $0$.

---

### Étape 2 : Retranscription mathématique

Pour traduire cette logique mathématique en une seule formule propre, on utilise l'opérateur "maximum" (ou la notation de la partie positive) :

$$\Phi(S_T) = \max(S_T - K, 0)$$

Que l'on note souvent de manière plus compacte :

$$\Phi(S_T) = (S_T - K)^+\$$

où l'exposant $^+$ signifie $\max(\text{expression}, 0)$.

---

### Étape 3 : Le prix du Call et l'utilisation de l'égalité en loi de $S_T$

Nous avons défini le payoff de notre Call européen à l'échéance $T$ :

$$\Phi(S_T) = (S_T - K)^+$$

Pour obtenir sa valeur aujourd'hui ($t = 0$), on applique le facteur d'actualisation $e^{-rT}$ et on prend l'espérance mathématique de ce payoff :

$$C_0 = e^{-rT} E\left[ (S_T - K)^+ \right]$$

Or, d'après notre **Chapitre 1**, nous avons établi que le prix de l'actif à l'échéance $S_T$ possède l'égalité en distribution suivante (grâce au Théorème de la Limite Centrale) :

$$S_T \overset{\mathcal{L}}{=} S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} Z \right)$$

où $Z$ est une variable aléatoire suivant la loi normale centrée réduite $\mathcal{N}(0, 1)$.

Puisque $S_T$ est défini via cette loi normale, calculer son espérance revient à intégrer sa fonction sur toute la loi de $Z$. Sa fonction de densité de probabilité étant $\varphi(z) = \frac{1}{\sqrt{2\pi}} e^{-\frac{z^2}{2}}$, notre espérance se réécrit sous forme d'intégrale, via le théorème de transfert :

$$E\left[ (S_T - K)^+ \right] = \int_{-\infty}^{+\infty} \left( S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) - K \right)^+ \varphi(z) \, dz$$

Et donc, pour le prix du Call $C_0$ :

$$C_0 = e^{-rT} \int_{-\infty}^{+\infty} \left( S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) - K \right)^+ \varphi(z) \, dz$$

Passons donc à l'étape suivante : **la gestion de la condition du $\max$ (le symbole $^+$) pour simplifier les bornes de notre intégrale**.

---

### Étape 4 : Simplification de l'intégrale grâce à la condition d'exercice

Notre intégrale actuelle est :

$$C_0 = e^{-rT} \int_{-\infty}^{+\infty} \left( S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) - K \right)^+ \varphi(z) \, dz$$

Le symbole $^+$ signifie que l'expression entre parenthèses n'est prise en compte que si elle est strictement positive. Sinon, elle vaut $0$ et l'intégrale n'apporte rien.

#### 1. Trouver la condition sur $z$ pour exercer l'option

L'option est exercée si et seulement si le prix de l'actif à l'échéance est supérieur au strike, c'est-à-dire $S_T > K$. En remplaçant $S_T$ par son expression avec $z$ :

$$S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) > K$$

Résolvons cette inéquation pour isoler la variable $z$ :

1. On divise par $S_0 > 0$ :

$$\exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) > \frac{K}{S_0}$$

2. On applique le logarithme népérien $\ln(\cdot)$ des deux côtés :

$$\left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z > \ln\left(\frac{K}{S_0}\right)$$

3. On isole le terme en $z$ :

$$\sigma \sqrt{T} z > \ln\left(\frac{K}{S_0}\right) - \left(r - \frac{\sigma^2}{2}\right)T$$

4. En remarquant que $\ln\left(\frac{K}{S_0}\right) = -\ln\left(\frac{S_0}{K}\right)$, on divise par $\sigma \sqrt{T}$ :

$$z > -\frac{\ln\left(\frac{S_0}{K}\right) + \left(r - \frac{\sigma^2}{2}\right)T}{\sigma \sqrt{T}}$$

#### 2. Définition du seuil $-d_2$

Ce terme barbare de droite est précisément ce que l'on appelle **$-d_2$** ! Posons donc :

$$d_2 = \frac{\ln\left(\frac{S_0}{K}\right) + \left(r - \frac{\sigma^2}{2}\right)T}{\sigma \sqrt{T}}$$

La condition $S_T > K$ se réécrit donc simplement : **$z > -d_2$**.

#### 3. Nouvelle écriture de l'intégrale

Puisque le gain est nul pour $z \le -d_2$ et vaut $S_T(z) - K$ pour $z > -d_2$, notre intégrale de $-\infty$ à $+\infty$ se simplifie en une intégrale bornée :

$$C_0 = e^{-rT} \int_{-d_2}^{+\infty} \left( S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) - K \right) \varphi(z) \, dz$$

---

### Étape 5 : Séparer l'intégrale en deux morceaux

Reprenons notre expression actuelle de $C_0$ :

$$C_0 = e^{-rT} \int_{-d_2}^{+\infty} \left[ S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) - K \right] \varphi(z) \, dz$$

Par simple distributivité, on peut séparer cette grande intégrale en deux intégrales distinctes :

$$C_0 = \underbrace{e^{-rT} \int_{-d_2}^{+\infty} S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) \varphi(z) \, dz}_{\text{Premier morceau (lié au prix de l'actif } S_0)} - \underbrace{e^{-rT} \int_{-d_2}^{+\infty} K \varphi(z) \, dz}_{\text{Second morceau (lié au strike } K)}$$

Attaquons-nous d'abord au **second morceau**, qui est le plus simple des deux !

---

### Le second morceau : L'apparition directe de $N(d_2)$

Regardons le second terme :

$$C_0^{(2)} =e^{-rT} K \int_{-d_2}^{+\infty} \varphi(z) \, dz$$

On sait que $\varphi(z)$ est la densité de probabilité d'une variable normale centrée réduite $Z \sim \mathcal{N}(0,1)$. Intégrer cette densité de $-d_2$ à $+\infty$, cela correspond exactement à la probabilité que $Z$ soit supérieur à $-d_2$ :

$$\int_{-d_2}^{+\infty} \varphi(z) \, dz = P(Z > -d_2)$$

Par symétrie de la loi normale, la probabilité que $Z > -d_2$ est égale à la probabilité que $Z \le d_2$. Et cette probabilité n'est autre que la fonction de répartition de la loi normale standard, que l'on note **$N(d_2)$** :

$$P(Z > -d_2) = P(Z \le d_2) = N(d_2)$$

Notre second morceau devient donc simplement :

$$C_0^{(2)} =e^{-rT} K N(d_2)$$

Et voilà ! La moitié de la formule de Black-Scholes est déjà là.

---

### Le premier morceau et l'apparition directe de $N(d_1)$

Reprenons le premier morceau de notre intégrale :

$$C_0^{(1)} = e^{-rT} \int_{-d_2}^{+\infty} S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) \varphi(z) \, dz$$

#### 1. Sortir et simplifier les constantes

Sortons d'abord tous les termes qui ne dépendent pas de $z$ ($S_0$, $e^{-rT}$ et $\exp\left( \left(r - \frac{\sigma^2}{2}\right)T \right)$) :

$$= S_0 \times e^{-rT} \times \exp\left( \left(r - \frac{\sigma^2}{2}\right)T \right) \int_{-d_2}^{+\infty} \exp\left( \sigma \sqrt{T} z \right) \varphi(z) \, dz$$

Simplifions le produit des deux termes en $T$ situés devant :

$$e^{-rT} \times \exp\left( \left(r - \frac{\sigma^2}{2}\right)T \right) = \exp\left( -rT + rT - \frac{\sigma^2}{2}T \right) = \exp\left( -\frac{\sigma^2}{2}T \right)$$

Il nous reste :

$$= S_0 \exp\left( -\frac{\sigma^2}{2}T \right) \int_{-d_2}^{+\infty} \exp\left( \sigma \sqrt{T} z \right) \varphi(z) \, dz$$

#### 2. Remplacer $\varphi(z)$ et transformer l'exposant

Remplaçons $\varphi(z)$ par sa formule explicite de densité normale centrée réduite ($\varphi(z) = \frac{1}{\sqrt{2\pi}} e^{-\frac{z^2}{2}}$) :

$$= S_0 \exp\left( -\frac{\sigma^2}{2}T \right) \int_{-d_2}^{+\infty} \frac{1}{\sqrt{2\pi}} \exp\left( \sigma \sqrt{T} z \right) \times \exp\left( -\frac{z^2}{2} \right) dz$$

En regroupant les exponentielles à l'intérieur ($\exp(A) \times \exp(B) = \exp(A+B)$), l'exposant devient $\sigma \sqrt{T} z - \frac{z^2}{2}$.
Appliquons l'astuce de la **complétion du carré** sur cet exposant :

$$-\frac{1}{2} \left( z^2 - 2\sigma \sqrt{T} z \right) = -\frac{1}{2} \left( (z - \sigma \sqrt{T})^2 - \sigma^2 T \right) = -\frac{(z - \sigma \sqrt{T})^2}{2} + \frac{\sigma^2 T}{2}$$

En réinjectant cet exposant transformé, nous obtenons l'expression clé :

$$S_0 \exp\left( -\frac{\sigma^2}{2}T \right) \int_{-d_2}^{+\infty} \frac{1}{\sqrt{2\pi}} \exp\left( -\frac{(z - \sigma \sqrt{T})^2}{2} + \frac{\sigma^2 T}{2} \right) dz$$

#### 3. Sortir le terme constant de l'intégrale

Le terme $\exp\left( \frac{\sigma^2 T}{2} \right)$ ne dépend pas de $z$ (c'est une constante par rapport à l'intégrale). On peut donc le sortir de l'intégrale et le multiplier avec le terme de devant :

$$S_0 \underbrace{\exp\left( -\frac{\sigma^2}{2}T \right) \exp\left( \frac{\sigma^2 T}{2} \right)}_{= \exp(0) = 1} \int_{-d_2}^{+\infty} \frac{1}{\sqrt{2\pi}} \exp\left( -\frac{(z - \sigma \sqrt{T})^2}{2} \right) dz$$

Il ne nous reste plus que :

$$S_0 \int_{-d_2}^{+\infty} \frac{1}{\sqrt{2\pi}} \exp\left( -\frac{(z - \sigma \sqrt{T})^2}{2} \right) dz$$

#### 4. Le changement de variable (L'astuce finale)

À l'intérieur de l'intégrale, nous avons un terme $(z - \sigma \sqrt{T})$ qui nous empêche d'avoir une simple densité normale centrée en $z$.

Faisons un **changement de variable** propre. Posons :

$$u = z - \sigma \sqrt{T}$$

Ce qui implique que :

* La différentielle est $du = 1 \times dz$
* Lorsque $z = -d_2$, la nouvelle borne $u$ devient : $-d_2 - \sigma \sqrt{T}$

Réécrivons cette borne en factorisant par le signe moins :

$$-\left( d_2 + \sigma \sqrt{T} \right)$$

C'est ici que l'on définit **$d_1$** ! On pose par convention :

$$d_1 = d_2 + \sigma \sqrt{T}$$

La nouvelle borne inférieure devient donc tout simplement **$-d_1$**.

#### 5. Obtenir $S_0 N(d_1)$

En remplaçant $z$ par $u$ dans notre intégrale, elle devient :

$$S_0 \int_{-d_1}^{+\infty} \frac{1}{\sqrt{2\pi}} e^{-\frac{u^2}{2}} du$$

L'expression $\frac{1}{\sqrt{2\pi}} e^{-\frac{u^2}{2}}$ n'est autre que la densité de probabilité de la loi normale centrée réduite $\varphi(u)$ !
Intégrer cette densité de $-d_1$ à $+\infty$, cela correspond à la probabilité qu'une variable normale standard $U$ soit supérieure à $-d_1$ :

$$\int_{-d_1}^{+\infty} \varphi(u) \, du = P(U > -d_1) = P(U \le d_1) = N(d_1)$$

Notre premier morceau donne donc finalement :

$$S_0 N(d_1)$$

---

### Résultat final: Assembler les deux morceaux

Maintenant, il ne reste plus qu'à réunir notre premier morceau ($S_0 N(d_1)$) et notre second morceau ($e^{-rT} K N(d_2)$) que nous avions calculés à l'étape précédente !

On obtient la célèbre formule du **Call européen de Black-Scholes** :

$$C_0 = S_0 N(d_1) - e^{-rT} K N(d_2)$$

## Le cas du Put Européen.

### Étape 1 : Le Payoff du Put Européen

Un **Put européen** confère à son détenteur le *droit* (mais non l'obligation) de **vendre** un actif sous-jacent à un prix d'exercice fixe (*strike*) $K$, à une date précise d'échéance $T$.

À l'échéance $T$, selon la valeur du prix de l'actif $S_T$ :

* **Cas 1 : L'actif vaut moins que le strike ($S_T < K$)**
On exerce l'option : on achète l'actif au prix du marché $S_T$ pour le revendre immédiatement au prix garanti $K$. Notre gain net est de :

$$K - S_T$$

* **Cas 2 : L'actif vaut plus ou autant que le strike ($S_T \ge K$)**
Il est plus intéressant de vendre l'actif directement sur le marché libre au prix $S_T$ plutôt qu'à $K$. On laisse l'option expirer. Le gain est de :

$$0$$

En résumant cela mathématiquement via la fonction de la partie positive, le payoff du Put à l'échéance $T$ s'écrit :

$$\Phi(S_T) = (K - S_T)^+$$

---

### Étape 2 : L'espérance mathématique et le prix initial

Pour ramener ce gain futur à aujourd'hui ($t = 0$), on applique le facteur d'actualisation continu $e^{-rT}$ et on prend l'espérance mathématique :

$$P_0 = e^{-rT} E\left[ (K - S_T)^+ \right]$$

En remplaçant $S_T$ par son expression continue via la loi normale standard $Z \sim \mathcal{N}(0,1)$ :

$$S_T \overset{\mathcal{L}}{=} S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} Z \right)$$

L'espérance se traduit sous forme d'intégrale en utilisant la densité de probabilité $\varphi(z) = \frac{1}{\sqrt{2\pi}} e^{-\frac{z^2}{2}}$ :

$$P_0 = e^{-rT} \int_{-\infty}^{+\infty} \left( K - S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) \right)^+ \varphi(z) \, dz$$

---

### Étape 3 : La condition d'exercice et les bornes de l'intégrale

Le Put n'exerce son effet que si le gain est strictement positif, c'est-à-dire si $K - S_T > 0, soit \)S_T < K$.

En remplaçant $S_T$ et en résolvant l'inéquation pour isoler $z$ (exactement comme pour le Call, mais inversée) :

$$S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) < K$$

$$z < -\frac{\ln\left(\frac{S_0}{K}\right) + \left(r - \frac{\sigma^2}{2}\right)T}{\sigma \sqrt{T}}$$

En reconnaissant notre terme **$d_2$** :

$$d_2 = \frac{\ln\left(\frac{S_0}{K}\right) + \left(r - \frac{\sigma^2}{2}\right)T}{\sigma \sqrt{T}}$$

La condition $S_T < K$ se réécrit simplement : **$z < -d_2$**.

Notre intégrale se borne donc de $-\infty$ jusqu'à $-d_2$ :

$$P_0 = e^{-rT} \int_{-\infty}^{-d_2} \left( K - S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) \right) \varphi(z) \, dz$$

---

### Étape 4 : Séparer l'intégrale en deux morceaux

Par distributivité, on sépare l'expression en deux intégrales distinctes :

$$P_0 = \underbrace{e^{-rT} K \int_{-\infty}^{-d_2} \varphi(z) \, dz}_{\text{Premier morceau (lié au strike } K)} - \underbrace{e^{-rT} \int_{-\infty}^{-d_2} S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) \varphi(z) \, dz}_{\text{Second morceau (lié au prix de l'actif } S_0)}$$

---

### Étape 5 : Résolution du premier morceau ($e^{-rT} K N(-d_2)$)

Le premier terme est direct :

$$e^{-rT} K \int_{-\infty}^{-d_2} \varphi(z) \, dz$$

Intégrer la densité normale de $-\infty$ à $-d_2$ correspond directement à la probabilité qu'une variable normale standard $Z$ soit inférieure ou égale à $-d_2$, c'est-à-dire la fonction de répartition **$N(-d_2)$** :

$$\int_{-\infty}^{-d_2} \varphi(z) \, dz = P(Z \le -d_2) = N(-d_2)$$

Le premier morceau vaut donc :

$$e^{-rT} K N(-d_2)$$

---

### Étape 6 : Résolution du second morceau ($S_0 N(-d_1)$)

Reprenons le second morceau :

$$e^{-rT} \int_{-\infty}^{-d_2} S_0 \exp\left( \left(r - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} z \right) \varphi(z) \, dz$$

En sortant les constantes et en appliquant la même **complétion du carré** sur l'exposant que pour le Call, l'intégrale devient :

$$S_0 \int_{-\infty}^{-d_2} \frac{1}{\sqrt{2\pi}} \exp\left( -\frac{(z - \sigma \sqrt{T})^2}{2} \right) dz$$

Faisons le changement de variable **$u = z - \sigma \sqrt{T}$** ($du = dz$) :

* Lorsque la borne supérieure est $z = -d_2$, la nouvelle borne devient : $-d_2 - \sigma \sqrt{T} = -(d_2 + \sigma \sqrt{T}) = -d_1$ (où $d_1 = d_2 + \sigma \sqrt{T}$).

L'intégrale se réécrit alors :

$$S_0 \int_{-\infty}^{-d_1} \frac{1}{\sqrt{2\pi}} e^{-\frac{u^2}{2}} du$$

Cette intégrale de $-\infty$ à $-d_1$ de la densité normale représente la probabilité $P(U \le -d_1)$, c'est-à-dire **$N(-d_1)$**.

Le second morceau vaut donc :

$$S_0 N(-d_1)$$

---

### Le résultat final: La formule du Put de Black-Scholes

En assemblant nos deux morceaux, on obtient la formule fermée du **Put européen de Black-Scholes** :

$$P_0 = e^{-rT} K N(-d_2) - S_0 N(-d_1)$$



**Comme démontré lors du Chapitre 1, ces deux formules ne sont que des estimations dans le cas discret et des estimations parfaites dans le cas continu, cadre dans lequel la formule de Black-Scholes s'instaure.**
