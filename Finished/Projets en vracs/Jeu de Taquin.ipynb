```python
import numpy as np
import pandas as pd
import numpy.random as rd
```

## Création du jeu

```python
class taquin:
    def __init__(self, l,w):
        self.l = l
        self.w = w
        self.lw = l*w
        self.grid = self.rshuffle()
        self.gridp = pd.DataFrame(self.grid)
        
    def locnan(self):
        return np.argwhere(np.isnan(self.grid)).squeeze()
    def shuffle(self):
        x = np.append(np.arange(1, self.lw, dtype=np.float16), np.nan)
        rd.shuffle(x)
        self.grid = x.reshape(self.l, self.w).astype(np.float16)
        return self.grid
        
    def rshuffle(self, max_tries=1000):
        for _ in range(max_tries):
            x = np.append(np.arange(1, self.lw, dtype=np.float16), np.nan)
            rd.shuffle(x)
            self.grid = x.reshape(self.l, self.w).astype(np.float16)
            if self.issolve():
                return self.grid
        raise ValueError("Impossible de générer une grille solvable après 1000 essais")
    #Définition des capabilités de déplacement de l'élément manquant.
    def cang(self):
        lo =self.w-1
        la = self.w - 1
        if self.locnan()[1] - 1 >= 0:
            return True
        else:
            return False
    def cand(self):
        lo =self.w-1
        la = self.w - 1
        if self.locnan()[1] + 1 <= la:
            return True
        else:
            return False
    def canh(self):
        lo =self.w-1
        la = self.w - 1
        if self.locnan()[0] - 1 >= 0:
            return True
        else:
            return False
    def canb(self):
        lo =self.w-1
        la = self.w - 1
        if self.locnan()[0] + 1 <= lo:
            return True
        else:
            return False
    #Fonctions directionnelles.
    
    def permute(self,game,i,j):
        i1,j1 = i
        i2,j2 = j
        if not(0<=i1 < game.shape[0] and 0<= j1 < game.shape[1]):
            raise IndexError("pos1 hors limites")
        if not(0<=i2 < game.shape[0] and 0<= j1 < game.shape[1]):
            raise IndexError("pos2 hors limites")
        game[i1,j1],game[i2,j2] = game[i2,j2],game[i1,j1]
        
    def g(self):
        if self.cang():
            self.permute(self.grid,self.locnan(),self.locnan() - [0,1])
        else:
            print("Vous ne pouvez pas tourner à gauche")
        return self.grid

    def d(self):
        if self.cand():
            self.permute(self.grid,self.locnan(),self.locnan() + [0,1])
        else:
            print("Vous ne pouvez pas tourner à droite")
        return self.grid
        
    def h(self):
        if self.canh():
            self.permute(self.grid,self.locnan(),self.locnan() - [1,0])
        else:
            print("Vous ne pouvez pas tourner en haut")
        return self.grid
        
    def b(self):
        if self.canb():
            self.permute(self.grid,self.locnan(),self.locnan() +[1,0])
        else:
            print("Vous ne pouvez pas tourner en bas")
        return self.grid
    def m(self,n):
        # Dictionnaire avec plusieurs alias par fonction et renvoie le mouvement voulu.

    
        """
        Sélecteur de mouvement. Plusieurs alias peuvent appeler la même fonction :
        - gauche : 0, 'g', 'left', 'l'
        - droite : 1, 'd', 'right', 'r'
        - haut   : 2, 'h', 'up', 'u'
        - bas    : 3, 'b', 'down', 'v'
        """
        # Dictionnaire alias → fonction
        mouvements = {
            "0": self.g, "g": self.g, "left": self.g, "l": self.g,
            "1": self.d, "d": self.d, "right": self.d, "r": self.d,
            "2": self.h, "h": self.h, "up": self.h, "u": self.h,
            "3": self.b, "b": self.b, "down": self.b, "v": self.b
        }
    
        # Convertir n en str pour accepter chiffres et lettres
        key = str(n).lower()  # lowercase pour 'UP', 'Left', etc.
    
        # Appeler la fonction correspondante si l'alias existe
        func = mouvements.get(key)
        if func:
            return func()
        else:
            print("Alias inconnu :", n)
        
    #On cherche à savoir si le jeu est possible d'être résolu.
    def issolve(self):
        """
        Retourne True si le taquin actuel est solvable, False sinon.
        """
        # 1️⃣ Remplacer np.nan par 0 et aplatir la grille
        taquin_flat = np.nan_to_num(self.grid, nan=0).flatten().tolist()
        largeur = self.w
        inversions = 0
        tuiles = [t for t in taquin_flat if t != 0]

        # 2️⃣ Compter les inversions
        for i in range(len(tuiles)):
            for j in range(i+1, len(tuiles)):
                if tuiles[i] > tuiles[j]:
                    inversions += 1

        # 3️⃣ Largeur impaire
        if largeur % 2 == 1:
            return inversions % 2 == 0

        # 4️⃣ Largeur paire : ligne du vide depuis le bas
        index_vide = taquin_flat.index(0)
        r = self.l - (index_vide // largeur)  # ligne depuis le bas, 1-indexed

        return (inversions + r) % 2 == 1
```
